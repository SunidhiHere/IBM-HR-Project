<div style="background-color:orange; padding:5px; border-radius:2px">
<h3 style="color: black;">Univariate Analysis</h3>
</div>

```python
# A. Continuous Variable (Age)
plt.figure(figsize=(8, 5))
sns.histplot(df['Age'], kde=True, color='skyblue', bins=20)
plt.title('Univariate Analysis: Distribution of Age')
plt.xlabel('Age')
plt.ylabel('Count')
plt.tight_layout()
plt.savefig('univariate_age_distribution.png')
plt.show()
```

```python
# B. Categorical Variable (Department)
plt.figure(figsize=(8, 5))
# Sorting the bars for clarity
dept_counts = df['Department'].value_counts().sort_values(ascending=False)
sns.barplot(x=dept_counts.index, y=dept_counts.values, palette='viridis')
plt.title('Univariate Analysis: Count of Employees by Department')
plt.xlabel('Department')
plt.ylabel('Number of Employees')
plt.tight_layout()
plt.savefig('univariate_department_count.png')
plt.show()
```

<div style="background-color:orange; padding:5px; border-radius:2px">
<h3 style="color: black;">Bivariate Analysis</h3>
</div>

```python
# A. Categorical vs Continuous (Attrition vs Monthly Income)
plt.figure(figsize=(8, 5))
sns.boxplot(x='Attrition', y='MonthlyIncome', data=df, palette='Set2')
plt.title('Bivariate Analysis: Monthly Income by Attrition Status')
plt.xlabel('Attrition')
plt.ylabel('Monthly Income ($)')
plt.tight_layout()
plt.savefig('bivariate_income_vs_attrition.png')
plt.show()
```

```python
# B. Categorical vs Categorical (Attrition vs OverTime)
plt.figure(figsize=(8, 5))
sns.countplot(x='OverTime', hue='Attrition', data=df, palette='coolwarm')
plt.title('Bivariate Analysis: Attrition Rate by Overtime Work')
plt.xlabel('Works Overtime')
plt.ylabel('Count')
plt.tight_layout()
plt.savefig('bivariate_overtime_vs_attrition.png')
plt.show()
```

```python
# C. Continuous vs Continuous (Total Working Years vs Monthly Income)
plt.figure(figsize=(9, 6))
sns.scatterplot(x='TotalWorkingYears', y='MonthlyIncome', hue='Attrition', data=df, alpha=0.7, palette='tab10')
plt.title('Bivariate Analysis: Monthly Income vs Total Working Years')
plt.xlabel('Total Working Years')
plt.ylabel('Monthly Income ($)')
plt.tight_layout()
plt.savefig('bivariate_income_vs_experience.png')
plt.show()
```

<div style="background-color:orange; padding:5px; border-radius:2px">
<h3 style="color: black;">Multivariate Analysis</h3>
</div>

```python
# A. Correlation Matrix (Heatmap) for all numeric attributes
plt.figure(figsize=(14, 6))
numeric_cols = df.select_dtypes(include=[np.number])
corr_matrix = numeric_cols.corr()
sns.heatmap(corr_matrix, annot=True, cmap="coolwarm", fmt=".2f", linewidths=0.5)
plt.tight_layout()
plt.show()
```

```python
# Draw the heatmap focusing on the strength of correlations
sns.heatmap(corr_matrix, annot=False, cmap='RdBu_r', vmin=-1, vmax=1, linewidths=0.5)
plt.title('Multivariate Analysis: Feature Correlation Heatmap', fontsize=18)
plt.tight_layout()
plt.savefig('multivariate_correlation_heatmap.png')
plt.show()
```

```python
# B. Multi-variable relationship (Age vs Monthly Income conditioned by Job Level & Attrition)
# This allows us to see how age and income behave across different job levels and attrition flags.
plt.figure(figsize=(10, 6))
sns.scatterplot(
    x='Age',
    y='MonthlyIncome',
    hue='Attrition',
    size='JobLevel',
    sizes=(20, 200),
    data=df,
    alpha=0.6,
    palette='magma'
)
plt.title('Multivariate Analysis: Age vs Income (by Attrition & Job Level)')
plt.xlabel('Age')
plt.ylabel('Monthly Income ($)')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.savefig('multivariate_bubble_chart.png')
plt.show()
```
